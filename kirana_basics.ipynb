# Task 1: Cleaning Product Names

```python
item1 = " toor-DAL 1kg "
item2 = "SUGAR (loose)"
item3 = " aashirvaad atta-10KG"
item4 = "tata-salt 1kg "

clean_item1 = item1.strip().replace("-", " ").title()
clean_item2 = item2.strip().replace("-", " ").title()
clean_item3 = item3.strip().replace("-", " ").title()
clean_item4 = item4.strip().replace("-", " ").title()

print(clean_item1)
print(clean_item2)
print(clean_item3)
print(clean_item4)
```

# Task 2: Splitting and converting Product Data

```python
record = "Toor Dal|1|168|42"

data = record.split("|")

item = data[0]
pack_kg = int(data[1])
price_inr = int(data[2])
stock = int(data[3])

print(f"{item.upper()} - {pack_kg} {'KG'.lower()} pack, price ₹{price_inr}, stock {stock}.")
```

# Task 3: Extracting SKU Information

```python
sku = "KIR-2025-GUJ-00123"

year = sku[4:8]
state_code = sku[9:12]
serial_number = sku[-5:]

print("Year:", year)
print("State Code:", state_code)
print("Serial Number:", serial_number)

print("Reversed SKU:", sku[::-1])
```

# Task 4: Generating a Receipt

```python
items = [
    ("Toor Dal 1Kg", 1, 168.00),
    ("Sugar (Loose)", 2, 90.00),
    ("Aashirvaad Atta", 1, 420.00),
    ("Tata Salt 1Kg", 1, 28.00)
]

total = 0

print(f"{'Item':<15}{'Qty':6}{'Amount':>10}")
print("-" * 31)

for item, quantity, amount in items:
    print(f"{item:<15}{quantity:6}{amount:>10,.2f}")
    total += amount

print("-" * 31)

print("Total: ₹{:,.2f}".format(total))
```
