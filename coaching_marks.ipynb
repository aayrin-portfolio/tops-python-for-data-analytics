1.<p>Create 'coaching_marks.ipynb' for a fictional coaching class. Define `calculate_percentage(marks_obtained, max_marks)` that returns the percentage rounded to 2 decimals, and call it for 3 students. Hint: Use `return`, not `print()`, so the value can be reused later.</p>

```python
# Define calculate_percentage function

def calculate_percentage(marks_obtained, max_marks):
    percentage = (marks_obtained / max_marks) * 100
    return round(percentage, 2)


# Calculate percentage for 3 students

student1 = calculate_percentage(450, 500)
student2 = calculate_percentage(380, 500)
student3 = calculate_percentage(420, 500)

print("Student 1 Percentage:", student1, "%")
print("Student 2 Percentage:", student2, "%")
print("Student 3 Percentage:", student3, "%")
```

2.<p>Define `assign_grade(percentage)` returning 'A+' for 90 and above, 'A' from 80, 'B' from 70, 'C' from 60, 'D' from 50 and 'F' below that. Test it on 6 values that include the boundaries 90, 80 and 49.99 and print each result.</p>

```python
# Define assign_grade function

def assign_grade(percentage):

    if percentage >= 90:
        return "A+"

    elif percentage >= 80:
        return "A"

    elif percentage >= 70:
        return "B"

    elif percentage >= 60:
        return "C"

    elif percentage >= 50:
        return "D"

    else:
        return "F"


# Test the function on 6 values

test_values = [95, 90, 80, 75, 60, 49.99]

for percentage in test_values:
    grade = assign_grade(percentage)
    print(percentage, "->", grade)
```

3.<p>Define `calculate_fee(base_fee, months=6, discount_pct=0)` that returns the total fee after discount, rounded to 2 decimals. Call it once with positional arguments, once with keyword arguments, once using only `base_fee`, and once with a 10 percent discount, printing all four totals.</p>

```python
# Define calculate_fee function

def calculate_fee(base_fee, months=6, discount_pct=0):

    total_fee = base_fee * months

    discount = total_fee * discount_pct / 100

    final_fee = total_fee - discount

    return round(final_fee, 2)


# 1. Positional arguments

fee1 = calculate_fee(2000, 3, 5)
print("Fee 1:", fee1)


# 2. Keyword arguments

fee2 = calculate_fee(base_fee=2000, months=4, discount_pct=10)
print("Fee 2:", fee2)


# 3. Using only base_fee

fee3 = calculate_fee(2000)
print("Fee 3:", fee3)


# 4. Using 10 percent discount

fee4 = calculate_fee(2000, discount_pct=10)
print("Fee 4:", fee4)
```

4.<p>Finish by defining `build_report_line(name, marks, max_marks=100)` that uses your two earlier functions and returns a string like 'Name | 78.0% | B'. Store 6 students as a list of tuples, loop through them printing each line, and print the class average percentage. Run Restart & Run All.</p>

```python
# Define build_report_line function

def build_report_line(name, marks, max_marks=100):

    percentage = calculate_percentage(marks, max_marks)

    grade = assign_grade(percentage)

    return f"{name} | {percentage}% | {grade}"


# Store 6 students as a list of tuples

students = [
    ("Ayesha", 95),
    ("Rahul", 85),
    ("Zoya", 78),
    ("Aarav", 65),
    ("Sara", 52),
    ("Ali", 45)
]


# Initialize total percentage

total_percentage = 0


# Print student reports

print("STUDENT REPORT")
print("-----------------------------")

for name, marks in students:

    percentage = calculate_percentage(marks, 100)

    print(build_report_line(name, marks))

    total_percentage += percentage


# Calculate class average

class_average = round(total_percentage / len(students), 2)

print("-----------------------------")

print("Class Average:", class_average, "%")
```
