1.<p>Open 'tiffin_orders.ipynb' and build the dictionary menu_prices with 6 meals and prices in rupees. Add 2 new meals, raise 2 prices by 10 percent rounded to a whole number, remove one meal with `pop()`, and print the dictionary after each change. Hint: If you missed the earlier session, start a new notebook with just this dictionary.</p>

```python
# Create dictionary with 6 meals

menu_prices = {
    "Gujarati Thali": 120,
    "Punjabi Thali": 150,
    "Dal Rice": 80,
    "Paneer Paratha": 100,
    "Veg Pulao": 90,
    "Chole Bhature": 110
}

print("Original Menu:")
print(menu_prices)


# Add 2 new meals

menu_prices["Masala Khichdi"] = 85
menu_prices["Veg Biryani"] = 130

print("\nAfter adding 2 meals:")
print(menu_prices)


# Increase 2 prices by 10%

menu_prices["Gujarati Thali"] = round(menu_prices["Gujarati Thali"] * 1.10)
menu_prices["Punjabi Thali"] = round(menu_prices["Punjabi Thali"] * 1.10)

print("\nAfter increasing prices:")
print(menu_prices)


# Remove one meal using pop()

menu_prices.pop("Chole Bhature")

print("\nAfter removing one meal:")
print(menu_prices)
```

3.<p>Create two sets, monday_customers and tuesday_customers, with 8 customer IDs each and at least 3 in common. Print the union with `|`, the intersection with `&`, the Tuesday-only IDs with `-`, and the size of each result.</p>

```python
# Create Monday and Tuesday customer sets

monday_customers = {
    "C101", "C102", "C103", "C104",
    "C105", "C106", "C107", "C108"
}

tuesday_customers = {
    "C103", "C104", "C105", "C109",
    "C110", "C111", "C112", "C113"
}


# Union - All unique customers from both days

all_customers = monday_customers | tuesday_customers

print("Union:", all_customers)
print("Union Size:", len(all_customers))


# Intersection - Customers common to both days

repeat_customers = monday_customers & tuesday_customers

print("\nIntersection:", repeat_customers)
print("Intersection Size:", len(repeat_customers))


# Difference - Customers only on Tuesday

tuesday_only = tuesday_customers - monday_customers

print("\nTuesday Only:", tuesday_only)
print("Tuesday Only Size:", len(tuesday_only))
```

```python
# Create daily summary dictionary

daily_summary = {
    "monday_count": len(monday_customers),

    "tuesday_count": len(tuesday_customers),

    "repeat_customers": len(repeat_customers),

    "repeat_rate_pct": round(
        len(repeat_customers) / len(monday_customers) * 100, 1
    ),

    "new_tuesday": len(tuesday_only)
}


# Print dictionary key by key

for key, value in daily_summary.items():
    print(key, ":", value)
```
