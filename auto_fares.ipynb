1.<p>Open 'auto_fares.ipynb' and create a list trip_distances of 12 distances between 0.8 and 18.5 km.
Use a `for` loop with `enumerate(trip_distances, start=1)` to print 'Trip n: X km -> Rs Y' using your
day-time fare rule from the earlier session. Constraint: Do not define any function. Hint: If you missed
the earlier session, use a flat Rs 14 per km.</p>

```python
trip_distances = [1.2, 2.5, 3.8, 4.5, 5.0, 6.7,
                  7.2, 8.5, 10.0, 12.5, 15.0, 18.5]

for n, distance in enumerate(trip_distances, start=1):
    fare = round(distance * 14,2)
    print(f"Trip {n}: {distance} km -> Rs {fare}")
```

2.<p>Add a while loop that adds fares from your trips one by one, using an index variable, and stops as soon as earnings reach Rs 1,500 or the trips run out. Print how many trips were needed and the final earnings.</p>

```python
total_earnings = 0
index = 0
trips_needed = 0

while index < len(trip_distances) and total_earnings < 1500:

    distance = trip_distances[index]

    fare = round(distance * 14, 2)

    total_earnings += fare

    trips_needed += 1

    index += 1

print("Trips needed:", trips_needed)
print("Final earnings: Rs", round(total_earnings, 2))
```

3.<p>Edit the list so that 2 trips have distance 0 (cancelled) and one trip is above 20 km. Loop through the trips, use continue to skip cancelled trips, and use break at the first trip above 20 km, printing its position and distance.</p>

```python
# Edit the list

trip_distances = [1.2, 0, 3.8, 4.5, 0, 6.7,
                  7.2, 8.5, 10.0, 12.5, 15.0, 22.5]

# Loop through trips

for position, distance in enumerate(trip_distances, start=1):

    if distance == 0:
        continue

    if distance > 20:
        print("Trip above 20 km found!")
        print("Position:", position)
        print("Distance:", distance, "km")
        break
```

4.<p>Finish with a loop that counts Short (under 3 km), Medium (3 to 8 km) and Long (above 8 km) trips in three counter variables and adds the total km and total fare. Print a 3-row summary table plus the totals, then run Restart & Run All. </p>

```python
# Initialize counters

short_trips = 0
medium_trips = 0
long_trips = 0

total_km = 0
total_fare = 0

# Loop through all trips

for distance in trip_distances:

    if distance == 0:
        continue

    if distance < 3:
        short_trips += 1

    elif distance <= 8:
        medium_trips += 1

    else:
        long_trips += 1

    total_km += distance

    fare = round(distance * 14, 2)

    total_fare += fare

# Print summary table

print("Trip Category\tCount")
print("------------------------")
print("Short\t\t", short_trips)
print("Medium\t\t", medium_trips)
print("Long\t\t", long_trips)

print("------------------------")
print("Total KM:", round(total_km, 2))
print("Total Fare: Rs", round(total_fare, 2))
```
