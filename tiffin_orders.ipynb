1.<p>Create 'tiffin_orders.ipynb' for a fictional Ahmedabad tiffin service. Build a list order_ids with 
8 IDs from 'T101' to 'T108', then use `append()` to add 'T109', `extend()` to add ['T110', 'T111'] and 
`pop()` to remove the last item, printing the list and `len()` after each step.</p>


```python
order_ids = ["T101","T102","T103","T104","T105","T106","T107","T108"]
print("list:", order_ids)
print("Length:", len(order_ids))

order_ids.append("T109")
print("After append:", order_ids)
print("Length:",len(order_ids))

order_ids.extend(["T110","T111"])
print("After extend:", order_ids)
print("Length:", len(order_ids))

order_ids.pop(10)
print("After pop:", order_ids)
print("Length:", len(order_ids))
```

2.<p>Create a list delivery_minutes with 8 whole numbers between 18 and 55 (at least two above 45).
Print the fastest and slowest with `min()` and `max()`, the average rounded to 1 decimal, and the
3 slowest with `sorted(delivery_minutes, reverse=True)[:3]`. Constraint: Do not use any loop.</p>

```python
delivery_minutes = [18, 24, 32, 41, 47, 52, 36, 49]

print("Delivery minutes:", delivery_minutes)

print("Fastest delivery:", min(delivery_minutes))
print("Slowest delivery:", max(delivery_minutes))

average = round(sum(delivery_minutes) / len(delivery_minutes), 1)
print("Average delivery:", average)

print("3 slowest deliveries:", sorted(delivery_minutes, reverse=True)[:3])
```
