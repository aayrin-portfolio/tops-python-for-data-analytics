1.<p>Open 'kirana_basics.ipynb' and add a section that stores 5 supplier price strings in variables: "46.5"
"120", "8.25", " 64 " and "72.00". Convert each with float (), print each value with its type() before and after, and print the sum rounded to 2 decimals. Hint: If you did not do the earlier session, start a new notebook with this section only. </p>

```python
price1 = "46.5"
price2 = "120"
price3 = "8.25"
price4 = " 64 "
price5 = "72.00"

print("Before conversion:")

print(price1, type(price1))
print(price2, type(price2))
print(price3, type(price3))
print(price4, type(price4))
print(price5, type(price5))

price1 = float(price1)
price2 = float(price2)
price3 = float(price3)
price4 = float(price4)
price5 = float(price5)

print("\nAfter conversion:")

print(price1, type(price1))
print(price2, type(price2))
print(price3, type(price3))
print(price4, type(price4))
print(price5, type(price5))

total = price1 + price2 + price3 + price4 + price5

print("\nTotal:", round(total, 2))
```

4.<p>Finish the section with variables for one sale line: item_name (str), price_in (float), quantity_kg (int) and is_organic (bool), each created by casting from a string you typed. Print a 4-line table of name, value and type (), and the line total as a float with 2 decimals. Run Restart & Run All before submitting.</p>

```python
item_name = str("Toor Dal")
price_inr = float("168.50")
quantity_kg = int("2")
is_organic = bool("True")

print("Name\t\tValue\t\tType")
print("-" * 45)

print("item_name\t", item_name, "\t", type(item_name))
print("price_inr\t", price_inr, "\t\t", type(price_inr))
print("quantity_kg\t", quantity_kg, "\t\t", type(quantity_kg))
print("is_organic\t", is_organic, "\t\t", type(is_organic))

line_total = float(price_inr * quantity_kg)

print("\nLine Total: ₹{:.2f}".format(line_total))
```
